```python
import os

print("🔍 Đang tìm kiếm thư mục 'root' phù hợp trong /kaggle/input...\n")

matching_roots = []

# Duyệt qua toàn bộ /kaggle/input
for root, dirs, files in os.walk("/kaggle/input"):
    # Kiểm tra nếu thư mục hiện tại chứa cả thư mục con 'images' và 'sparse'
    if "images" in dirs and "sparse" in dirs:
        matching_roots.append(root)

if matching_roots:
    print("✅ Đã tìm thấy đường dẫn root thích hợp:")
    for path in matching_roots:
        print("-" * 50)
        print(f'dataset:\n  root: "{path}"')
        print("-" * 50)
else:
    print("⚠️ Không tìm thấy thư mục nào chứa đồng thời cả 'images' và 'sparse'.")
    print("Dưới đây là toàn bộ cấu trúc thư mục trong /kaggle/input để bạn kiểm tra:\n")
    for root, dirs, files in os.walk("/kaggle/input"):
        print(root)
```
